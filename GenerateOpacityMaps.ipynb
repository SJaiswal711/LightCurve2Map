```python
from hashlib import sha256
import os
```

```python
def augment_unique_with_hash_exclusion(opacity_maps, global_seen=None):
    """
    Augments and filters for global uniqueness using a shared hash set.

    Parameters
    ----------
    opacity_maps : np.ndarray
        Binary opacity maps (N, H, W).
    global_seen : set
        Set of previously seen hashes (for leakage prevention).

    Returns
    -------
    unique_augmented_maps : np.ndarray
        Globally unique augmented maps.
    unique_counts : np.ndarray
        Number of unique augmentations per input shape.
    global_seen : set
        Updated set of global hashes.
    """
    all_unique = []
    unique_counts = []
    if global_seen is None:
        global_seen = set()

    for img in opacity_maps:
        variations = [
            img,
            np.flipud(img),
            np.fliplr(img),
            np.rot90(img, 1),
            np.rot90(img, 2),
            np.rot90(img, 3),
            np.fliplr(np.rot90(img, 1)),
            np.flipud(np.rot90(img, 1))
        ]

        count = 0
        for var in variations:
            h = sha256(var.tobytes()).hexdigest()
            if h not in global_seen:
                global_seen.add(h)
                all_unique.append(var)
                count += 1

        unique_counts.append(count)

    return np.array(all_unique), np.array(unique_counts), global_seen


def split_and_augment_strict(opacity_maps, test_ratio=0.025, val_ratio=0.025, seed=42):
    np.random.seed(seed)
    total = len(opacity_maps)
    indices = np.arange(total)
    np.random.shuffle(indices)

    test_size = int(total * test_ratio)
    val_size = int(total * val_ratio)

    test_idx = indices[:test_size]
    val_idx = indices[test_size:test_size+val_size]
    train_idx = indices[test_size+val_size:]

    test_maps = opacity_maps[test_idx]
    val_maps = opacity_maps[val_idx]
    train_maps = opacity_maps[train_idx]

    # Augment test → update global hash set
    test_aug, test_counts, global_seen = augment_unique_with_hash_exclusion(test_maps)

    # Augment val → skip those already in test
    val_aug, val_counts, global_seen = augment_unique_with_hash_exclusion(val_maps, global_seen)

    # Augment train → skip anything in test/val
    train_aug, train_counts, _ = augment_unique_with_hash_exclusion(train_maps, global_seen)

    return {
        "train": (train_maps, train_aug, train_counts),
        "val": (val_maps, val_aug, val_counts),
        "test": (test_maps, test_aug, test_counts)
    }
```

```python
def save_augmented_splits(split_dict, name_prefix, save_dir="data"):
    """
    Save augmented opacity map splits to disk.

    Parameters
    ----------
    split_dict : dict
        Output of split_and_augment_strict(). Contains keys 'train', 'val', and 'test'.
        Each value is a tuple: (original_maps, augmented_maps, augmentation_counts)
    name_prefix : str
        Prefix to use for all filenames (e.g. "typeIa" → "typeIa_train_OM.npy", etc.)
    save_dir : str
        Directory where the files will be saved. Created if it doesn't exist.
    """

    os.makedirs(save_dir, exist_ok=True)

    for split_name in ['train', 'val', 'test']:
        orig, aug, counts = split_dict[split_name]

        np.save(os.path.join(save_dir, f"Ori_{split_name}_{name_prefix}.npy"), orig)
        np.save(os.path.join(save_dir, f"Aug_{split_name}_{name_prefix}.npy"), aug)
        np.save(os.path.join(save_dir, f"Num_{split_name}_{name_prefix}.npy"), counts)

        print(f"[✓] Saved {split_name} set: {len(orig)} original → {len(aug)} unique augmented maps")

```

```python
all_types = [typeIa, typeIb, typeIIa, typeIIb, typeIII, typeIV, typeV, typeVI]
type_names = ["typeIa", "typeIb", "typeIIa", "typeIIb", "typeIII", "typeIV", "typeV","typeVI"]

# Loop through each type, split, augment, and save
for data, name in zip(all_types, type_names):
    print(f"\n🚀 Processing: {name}")
    splits = split_and_augment_strict(data)  # Perform split & augment
    save_augmented_splits(splits, name_prefix=name, save_dir="Data/OM")  # Save to disk
```
